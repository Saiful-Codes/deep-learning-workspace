```python
import os
from collections import Counter
from PIL import Image
import explore
# Handle module reloading
importlib.reload(explore)

# --- Convert one-hot labels to integer class indices ---
# Column 0 is the image id; columns 1..7 are the one-hot class columns.
# argmax(axis=1) returns the position of the single "1" in each row,
# which IS the integer class index (matching IMG_CLASS_NAMES order).
train_labels = train_df.iloc[:, 1:].values.argmax(axis=1)
val_labels = val_df.iloc[:, 1:].values.argmax(axis=1)

# --- Sanity check: integer labels vs. the raw one-hot rows ---
print("First 20 train labels:", train_labels[:20])
display(train_df.head(20))   # compare each one-hot row to the label above

# --- Class distribution plots (mandatory check) ---
explore.plot_label_distribution(train_labels, 'train', IMG_CLASS_NAMES)
explore.plot_label_distribution(val_labels, 'val', IMG_CLASS_NAMES)

# --- Additional data check: image integrity (missing files + size consistency) ---
# Open every referenced image once (PIL reads only the header for .size, so it's fast).
# A missing file would crash training; inconsistent sizes would break a fixed Resize.
missing = []
sizes = Counter()
for img_id in train_df['image']:
    path = f'data/img/{img_id}.jpg'
    if not os.path.exists(path):
        missing.append(img_id)
        continue
    sizes[Image.open(path).size] += 1

print(f"\nMissing image files referenced in train.csv: {len(missing)}")
print(f"Unique image sizes (W, H) -> count: {dict(sizes)}")
```

## Model Training Cell

We will be using Weights and Biases to keep track of our experimental runs and evaluation metrics. The Weights and Biases tool is only used in "Text Sentiment Analysis" lab. If you have not completed this lab yet, you can get started with Weights and Biases by following Steps 1a), b) and c) from the following link: https://docs.wandb.ai/quickstart


```python
import os
# Keeps W&B from blocking on network sync; train_model still calls wandb.init/log,
# so the logging structure and metric keys are unchanged (they just no-op).
# Remove / set to "online" once you want runs to appear in the W&B dashboard (Task 1f/1g).
os.environ["WANDB_MODE"] = "disabled"

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torch.utils.data.sampler import WeightedRandomSampler

import datasets
import models
import train

# Handle module reloading
importlib.reload(datasets)
importlib.reload(models)
importlib.reload(train)

torch.manual_seed(42)

NUM_EPOCHS = 5
BATCH_SIZE = 64

# Create datasets/loaders
# Task 1b - debug on the _small CSVs first, then switch to train.csv / val.csv.
# Task 1e - augment=True ONLY for training; validation stays deterministic.
train_dataset = datasets.LesionDataset('data/img', 'data/img/train.csv', augment=True)
val_dataset   = datasets.LesionDataset('data/img', 'data/img/val.csv',   augment=False)

# Task 1d - counter the class imbalance found in Task 1a with a WeightedRandomSampler.
# Inverse-frequency weights: a rare class (e.g. DF) gets a large per-sample weight and a
# common class (NV) a small one, so each epoch draws a roughly class-balanced stream.
class_counts = np.bincount(train_dataset.labels, minlength=len(IMG_CLASS_NAMES))
class_weights = 1.0 / class_counts
sample_weights = class_weights[train_dataset.labels]
sampler = WeightedRandomSampler(weights=sample_weights,
                                num_samples=len(train_dataset), replacement=True)
# Use the sampler instead of shuffle=True (the two are mutually exclusive).
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, sampler=sampler)
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False)

# Instantiate model, optimizer and criterion
# Task 1c - baseline 5-block batch-norm CNN.
model = models.SimpleBNConv(num_classes=len(IMG_CLASS_NAMES))

optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
# Task 1d - the sampler above already rebalances each batch, so a plain loss is kept
# (only one variable changes vs. the baseline). An alternative is a class-weighted loss
# instead of the sampler:
#   cw = torch.tensor(class_weights, dtype=torch.float32, device=train.device)
#   criterion = nn.CrossEntropyLoss(weight=cw)
criterion = nn.CrossEntropyLoss()

# Train model
# Task 1e - ident_str marks this as the weighted + augmented run.
train.train_model(model, train_loader, val_loader, optimizer, criterion,
                  IMG_CLASS_NAMES, NUM_EPOCHS, project_name="CSE3001 Assignment Task 1",
                  ident_str="weighted_aug_full")
```
