# Vectorisation

Why does this matter??

Let's take a look at how we can find the average value in a tensor

```python
import time
```

```python
# batch, height, width
tensor_x = torch.rand(4, 2, 3)
# tensor_x = torch.rand(40, 20, 30)
# tensor_x = torch.rand(400, 200, 300)
print(tensor_x)
print(tensor_x.shape)
```

## The inefficient way (non-vectorised)

```python
# For loops!
start = time.perf_counter()

sum = 0
for i in range(0, 400, 1):
  for j in range(0, 200, 1):
    for k in range(0, 300, 1):
      sum = sum + tensor_x[i,j,k]
avg = sum / (400 * 200* 300)
print(avg)


end = time.perf_counter()
print(f'Elapsed: {end-start:.3f}s')
```

## The better way (vectorised!)

```python
start = time.perf_counter()

# compute the mean across the entire vector
avg = tensor_x.mean()
print(avg)
print(tensor_x.shape)

# compute the mean across just the final dimension (dimension 2)
avg = tensor_x.mean(axis=(2,))
print(avg.shape)

end = time.perf_counter()
print(f'Elapsed: {end-start:.3f}s')
```

# Reshaping/Permuting

First do some examples of reshape versus permute on the board.

Understanding the difference between reshape and permute. Notice in the example below reshape keeps the order of the elements. Whereas permute modifies the order.


```python
tensor_x = torch.arange(24, dtype=torch.int)
print(tensor_x)

# lets make tensor_x into shape(2, 4, 3) (should we use reshape or permute)?
tensor_x = tensor_x.reshape(2, 4, 3)
print(tensor_x)

# lets combine the last two dimensions (should we use reshape or permute)?
tensor_x = tensor_x.reshape(2, 4*3)
print(tensor_x)

# lets get vector_x back to 1D vector (should we use reshape or permute)?
# notice the order.
tensor_x = tensor_x.reshape(24)
print(tensor_x)

# lets get tensor_x to shape (2, 12) (should we use reshape or permute)?
tensor_x = tensor_x.reshape(2, 12)
print(tensor_x)

# lets switch dimension 0 and dimension 1 (should we use reshape or permute)?
tensor_x = tensor_x.permute(1, 0)
print(tensor_x)

# lets reshape back to 1D
# notice the order
tensor_x = tensor_x.reshape(24)
print(tensor_x)

```

### Where is reshape and permute used in practice?

Consider you are given a 4D tensor representing an image of shape(batch, channels, height, width)
1. What if a model requires the tensor representing the image to be in the following shape instead shape(batch, height, width, channels).
2. What if we need to flatten the tensor so that we use a 1D vector to represent each example in the batch? Similar transformations is often need in practice in when training convolutional neural networks.

```python
# You are given the following tensor for an image
tensor_image = torch.rand(25, 3, 255, 255)

# We use permute to move the channel to the last dimension
tensor_image_modified = tensor_image.permute(0,2,3,1)
print(tensor_image_modified.shape)

# We use reshape to combine the final dimensions to flatten the image
tensor_image_flattened = tensor_image_modified.reshape(25, 3*255*255)
print(tensor_image_flattened.shape)
```
